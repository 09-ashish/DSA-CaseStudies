```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
mpg = sns.load_dataset("mpg")
```

```python
sns.regplot(data=mpg, x="displacement", y="mpg", scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title("Fuel Efficiency vs. Engine Displacement")
plt.xlabel("Engine Displacement (cu. in.)")
plt.ylabel("Miles Per Gallon (MPG)")
sns.despine()
plt.show()
```

As engine displacement increases, fuel efficiency decreases significantly.

```python
sns.boxplot(data=mpg, x="origin", y="mpg", hue="origin", palette="Set2", legend=False)
sns.stripplot(data=mpg, x="origin", y="mpg", color="black", alpha=0.3, jitter=0.2)
plt.title("Distribution of MPG by Vehicle Origin")
plt.xlabel("Country of Origin")
plt.ylabel("Miles Per Gallon (MPG)")
sns.despine()
plt.show()
```

Cars manufactured in Japan and Europe generally have much higher MPG values than cars manufactured in the USA.

```python
numeric_mpg = mpg.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_mpg.corr(), annot=True, cmap="mako", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Vehicle Attributes")
plt.show()
```

weight,cylinders and displacement have strong positive correlation. they also have strong negative correlation with mpg.


```python
sns.scatterplot(
    data=mpg,
    x="weight",
    y="acceleration",
    hue="cylinders",
    size="cylinders",
    palette="viridis",
    sizes=(40, 140),
    alpha=0.7
)
plt.title("Vehicle Weight vs. Acceleration Time")
plt.xlabel("Weight (lbs)")
plt.ylabel("Time to Accelerate 0-60 mph (seconds)")
sns.despine()
plt.legend(title="Cylinders")
plt.show()
```

Lighter vehicles take more time to accelarate from 0-60 mph
